```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
avg_close = df.groupby('Year')['Close'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(x='Year', y='Close', data=avg_close, palette='plasma')
plt.title('Average Closing Price by Year')
plt.show()
```

```python
monthly_avg = df.groupby('Day')['Close'].mean().sort_values()
plt.figure(figsize=(10,6))
sns.barplot(x=monthly_avg.index, y=monthly_avg.values, palette='viridis')
plt.title("Distribution of Daily Returns")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Daily Return'], bins=50, kde=True, color='teal')
plt.title("Distribution of Daily Returns")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Volume'], fill=True, color='purple')
plt.title('Volume Density')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Open'], fill=True, color='red')
plt.title('Volume Density')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='Day', y='Close', palette='Spectral')
plt.title("Close Price Distribution by Month")
plt.show()
```

```python
plt.figure(figsize=(12,6))
df['Open'].rolling(window=30).mean().plot(color='green')
plt.title('30-Day Rolling Mean of Close Price')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['Open', 'Close', 'High', 'Low']], palette='coolwarm')
plt.title("Spread of Stock Prices")
plt.show()
```

```python
df[['Open','Close','High','Low']].plot.area(figsize=(12,6), alpha=0.5, cmap='coolwarm')
plt.title("Stock Price Components Over Time")
plt.xlabel("Index / Date")
plt.ylabel("Price")
plt.show()

```

```python
df['Year'] = df['Date'].dt.year

plt.figure(figsize=(10,6))
sns.violinplot(x='Year', y='Volume', data=df, palette='Set2')
plt.title("Volume Distribution by Year")
plt.show()
```
